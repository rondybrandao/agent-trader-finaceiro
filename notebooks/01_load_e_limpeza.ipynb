```python
import urllib.request
import zipfile
from pathlib import Path
import os

# Caminho pra data/raw
RAW = Path.cwd() / 'data' / 'raw'
if not RAW.exists():
    RAW = Path.cwd().parent / 'data' / 'raw'
RAW.mkdir(parents=True, exist_ok=True)
print(f"Salvando em: {RAW}")

# Anos que queremos
ANOS = [2020, 2021, 2022, 2023]

for ano in ANOS:
    zip_url = f"http://bvmf.bmfbovespa.com.br/InstDados/SerHist/COTAHIST_A{ano}.ZIP"
    zip_path = RAW / f"COTAHIST_A{ano}.ZIP"
    txt_path = RAW / f"COTAHIST_A{ano}.txt"
    
    # Pula se já existe o TXT
    if txt_path.exists():
        print(f"✅ {ano} já baixado e extraído")
        continue
    
    # Baixa o ZIP
    print(f"\n📥 Baixando {ano}...")
    try:
        urllib.request.urlretrieve(zip_url, zip_path)
        print(f"  ZIP salvo: {zip_path} ({zip_path.stat().st_size / 1e6:.1f} MB)")
    except Exception as e:
        print(f"  ❌ Erro ao baixar {ano}: {e}")
        continue
    
    # Extrai
    print(f"📦 Extraindo {ano}...")
    try:
        with zipfile.ZipFile(zip_path, 'r') as z:
            z.extractall(RAW)
        print(f"  ✅ Extraído")
    except Exception as e:
        print(f"  ❌ Erro ao extrair: {e}")
        continue
    
    # Remove o ZIP pra economizar espaço
    zip_path.unlink()
    print(f"  🗑️ ZIP removido")

# Lista o que temos
print("\n=== Arquivos em data/raw ===")
for f in sorted(RAW.glob('COTAHIST*')):
    print(f"  {f.name}  ({f.stat().st_size / 1e6:.0f} MB)")
```
